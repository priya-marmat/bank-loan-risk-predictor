```python
def cap_upper_outliers(df, columns):
    df_clean = df.copy()
    for col in columns:
        Q3 = df_clean[col].quantile(0.75)
        IQR = df_clean[col].quantile(0.75) - df_clean[col].quantile(0.25)
        upper = Q3 + 1.5 * IQR
        
        before = df_clean[col].max()
        df_clean[col] = df_clean[col].clip(upper=upper)  # only cap upper!
        after = df_clean[col].max()
        
        print(f"✅ {col}: max changed from {before:,.0f} → {after:,.0f}")
    
    return df_clean

problem_cols = ['annual_income', 'loan_amount', 'loan_installment', 'goods_price']
application_df = cap_upper_outliers(application_df, problem_cols)
```

```python
num_cols = ['annuity_amount', 'applied_amount', 'approved_amount']

def cap_upper_outliers(df, columns):
    df_clean = df.copy()
    for col in columns:
        Q3 = df_clean[col].quantile(0.75)
        Q1 = df_clean[col].quantile(0.25)
        IQR = Q3 - Q1
        upper = Q3 + 1.5 * IQR
        
        before = df_clean[col].max()
        df_clean[col] = df_clean[col].clip(upper=upper)
        after = df_clean[col].max()
        
        print(f"✅ {col}: max changed from {before:,.0f} → {after:,.0f}")
    
    return df_clean

previous_df = cap_upper_outliers(previous_df, num_cols)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

for i, col in enumerate(num_cols):
    snc.boxplot(y=previous_df[col], ax=axes[i], color='lightgreen')
    axes[i].set_title(col, fontweight='bold')

plt.suptitle('✅ After Capping — Previous Loans', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
